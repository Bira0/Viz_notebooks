```python
# Import libraries
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import chi2_contingency
from statsmodels.graphics.mosaicplot import mosaic
import statsmodels.formula.api as smf

# Optional: make plots look clear
sns.set_theme(style="whitegrid")
```

## 1. Read the data

In the R file, the dataset was already saved in an object called `loans_data`.

In Python, you can read the data from a CSV file. Replace the file name below with your actual file name.

```python
# Option 1: read from CSV file
loans = pd.read_csv("/kaggle/input/datasets/eimieimieimi86/bank-data/financial_loan_modified.csv")

# Option 2: if you already have a DataFrame called loans_data, use:
# loans = loans_data.copy()

# For now, uncomment one of the lines above before running the notebook.
# Display first rows after loading:
loans.head()
```

## 2. Check the categorical variables

We need two categorical variables:

- `homeownership`
- `application_type`

```python
# Check column names
# loans.columns

# Check unique values
print(loans["home_ownership"].unique())
print(loans["application_type"].unique())
```

## 3. Bar chart: Homeownership counts

This plot shows how many observations are in each `homeownership` category.

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=loans, x="home_ownership")
plt.xlabel("Homeownership")
plt.ylabel("Count")
plt.title("Count of Homeownership Categories")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## 4. Grouped bar chart

This is similar to `fill = application_type` in ggplot2. It shows counts of `homeownership`, separated by `application_type`.

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=loans, x="home_ownership", hue="application_type")
plt.xlabel("Homeownership")
plt.ylabel("Count")
plt.title("Homeownership by Application Type")
plt.xticks(rotation=30)
plt.legend(title="Application type")
plt.tight_layout()
plt.show()
```

## 8. Contingency table

A contingency table shows the number of observations for each combination of the two categorical variables.

```python
tab = pd.crosstab(loans["home_ownership"], loans["application_type"])
tab
```

## 9. Mosaic plot

A mosaic plot is useful for showing the relationship between two categorical variables.

```python
plt.figure(figsize=(8, 5))
mosaic(loans, ["home_ownership", "application_type"],labelizer=lambda k: '')
plt.title("Homeownership vs Application Type")
plt.tight_layout()
plt.show()
```

## 10. Chi-square test

The chi-square test checks whether there is a significant association between two categorical variables.

### Hypotheses

**H0:** `homeownership` and `application_type` are independent (No association)

**H1:** `homeownership` and `application_type` are associated (association).

```python
chi2, p_value, dof, expected = chi2_contingency(tab)

print("Chi-square statistic:", chi2)
print("Degrees of freedom:", dof)
print("p-value:", p_value)

if p_value < 0.05:
    print("Conclusion: Reject H0. There is a significant association.")
else:
    print("Conclusion: Fail to reject H0. No significant association was found.")
```

## 13. Pie chart: Distribution of homeownership

A pie chart can show the distribution of one categorical variable. However, for many categories, a bar chart is usually easier to read.

```python
df_counts = loans["home_ownership"].value_counts()

df_counts
```

```python
plt.figure(figsize=(6, 6))
plt.pie(
    df_counts,
    labels=df_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Distribution of Homeownership")
plt.tight_layout()
plt.show()
```
